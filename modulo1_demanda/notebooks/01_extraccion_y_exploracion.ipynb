# Módulo 1 — Extracción y Exploración Inicial con Polars

Fuente: [Operación de pasajeros y despacho de vehículos en la modalidad de transporte de pasajeros por carretera](https://www.datos.gov.co/Transporte/Operaci-n-de-pasajeros-y-despacho-de-veh-culos-en-/eh75-8ah6/data_preview) (`eh75-8ah6`).

Los datos se almacenan en formato **Parquet** comprimido (`zstd`) en `modulo1_demanda/data/raw/operacion_pasajeros.parquet` (ignorado por `.gitignore`) para lectura ultrarrápida y procesamiento exclusivo con **Polars**.

```python
from pathlib import Path
import polars as pl

DATA_PATH = Path("../data/raw/operacion_pasajeros.parquet")
print(f"Polars: {pl.__version__} | Archivo existe: {DATA_PATH.exists()}")
```

```python
# Lectura lazy (sin cargar los ~49M de registros completos en RAM)
lf = pl.scan_parquet(DATA_PATH)
print("Esquema:")
print(lf.collect_schema())
```

```python
# Conteo total y rango de fechas
resumen = lf.select([
    pl.len().alias("total_filas"),
    pl.col("fecha_despacho").min().alias("fecha_min"),
    pl.col("fecha_despacho").max().alias("fecha_max"),
    pl.col("pasajeros").sum().alias("total_pasajeros"),
    pl.col("despachos").sum().alias("total_despachos"),
]).collect()

resumen
```
